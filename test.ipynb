```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
import ast
import seaborn as sns

# Loading Data
ds = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/esempio_analisi_dati_vendite.csv')
df = pd.DataFrame(ds)
```

```python
# Data Cleanup
# Specify the correct format
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
```

```python
df.head()
```

```python
df_plot = df['Region'].value_counts().to_frame().head(30)
```

```python
sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='Region', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Number of Products per Region')
plt.xlabel('Number of Prd')
plt.ylabel('')
plt.show()
```
